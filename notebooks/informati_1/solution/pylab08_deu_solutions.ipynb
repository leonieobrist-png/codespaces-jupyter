Diese Übung soll Euch helfen den Umgang mit Funktionen zu verstehen und zu üben.

In den weiteren Übungen werden wir Euch z.T. die funktion schriftlich definieren.
Solche Spezifikationen enthalten den Namen der Funktion, die Namen der Argumente (falls vorhanden), ihre Reihenfolge und den Rückgabewert der Funktion. 
Auch wenn sich die Namen der Argumente ändern können, sollte alles andere gleich bleiben. 
Hier ist ein Beispiel für eine Funktionsspezifikation.

* Name: `die_funktion`
* Argumente:
  * `arg1` (string) - Das erste Argument
  * `arg2` (Ganzzahl) - Das zweite Argument 
* Rückgabewerte:
  * Ihr Name (Zeichenkette)  

Die Antwort auf die Funktionsspezifikation wird folgendermassen aufgebaut:

```python
def the_function(arg1, arg2): 
    """Dies ist der docstring für the_function"""
    # Füge deinen Code hier ein
    return "Mike"
```

Bevor du deine Funktion einreichst oder testest, stelle sicher, dass:

1. Du hast den richtigen Namen verwendet
1. Es gibt einen Docstring
1. Du hast die gleiche Anzahl von Argumenten (du kannst sie umbenennen, wenn du willst)
1. Du gibst den angeforderten Wert zurück. Wird kein Rückgabewert spezifiziert, gibt die Funktion 'None' zurück.

### 6. Rückgabe eines bool' schen Wertes

Schreibe eine Funktion namens `is_greater`, die zwei Argumente namens `a` und `b` annimmt. Die Funktion gibt `True` zurück, wenn `a` größer als `b` ist, ansonsten `False`.

* Name: `is_greater`
* Argumente:
    * `a` (Float) Eine Zahl 
    * `b` (Float) Eine Zahl 
* Rückgabe:
    * Wahr, wenn `a` größer als `b` ist. Andernfalls `False`. 

````python
# Solution 1

def is_greater(a, b):
    return a > b

print(is_greater(1,3))
print(is_greater(3,1))
````

````python
# Solution 2

def is_greater(a, b):
    if a > b:
        return True
    else:
        return False
    
print(is_greater(1,3))
print(is_greater(3,1))
````
