## Part 1: Using Dictionaries

### 1. Create a Dictionary 

Create a dictionary called `courses` that contains the following:


| Key | Value | 
| --- | --- |
| Mat | Mathematics |
| Phy | Physics |
| Eng | English | 
| Inf | Informatics |



````python
courses = {
    'Mat': 'Mathematics',
    'Phy': 'Physics',
    'Eng': 'English',
    'Inf': 'Informatics',
}
````

### 1.1 Print all Keys

Use the `courses` dictionary you created. Write a `print` statement that prints all keys of the dictionary.

````python
print(courses.keys())
````

### 1.1 Print all Values

Use the `courses` dictionary you created. Write a `print` statement that prints all values of the dictionary.

````python
print(courses.values())
````

### 2. Print a Key

Use the `courses` dictionary you created. Write a `print` statement that prints the title of `Inf`.

````python
# Solution 1: accessing the value with key
print(courses['Inf'])
````

````python
# Solution 2: with get() method
print(courses.get('Inf'))
````

````python
# Solution 3: get values of dict as list and access index of list
values = courses.values()
# this will return a python object called 'dict_value'
# we can convert this object into a list:
values_as_list = list(values)
# and now we can access the items of the list by index:
print(values_as_list[0])

# as one-liner
print(list(courses.values())[0])
````

### 3. Change a Value 

The name of CIS-15 changed to "Python Programming for Everyone". Change the value in `courses` to reflect the new name.

````python
# Solution 1: accessing key and overwrite value
new_value = 'Informatik HS22'

courses['Inf'] = new_value
print(courses)
````

````python
# Solution 2: update method
courses.update({'Inf':new_value})
print(courses)
````

````python
# Solution 3: deleting and inserting
# this will change the order of the dict. 'Inf' is now the last entry.
del courses['Inf']
courses['Inf'] = new_value
print(courses)
````

### 5. Remove a Key 

Write a statement that removes `Phy` from the dictionary. 



*If you try to remove a non existing key, you will get an error massege!*

If you want to try all the following solutions, you must save the dictionary again at the beginning of this part .

````python
# Solution 1: pop() method
key_to_remove = 'Phy'
courses.pop(key_to_remove)
print(courses)
````

````python
# Solution 2: del operator
del courses[key_to_remove]
print(courses)
````

````python
# optional: an exapmle to remove multiple keys
keys_to_remove = ['Eng', 'MedMi']

for key in keys_to_remove:
    courses.pop(key)
    
print(courses)
````
