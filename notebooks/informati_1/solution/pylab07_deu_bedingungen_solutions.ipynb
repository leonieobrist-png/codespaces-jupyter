### 3. Verbesserte Buchstabennote 

Verbessere deine Funktion `grade_letter_percent` so, dass sie den Prozentsatz auf der Grundlage der Gesamtpunktzahl einer Person und der möglichen Punkte für eine Aufgabe berechnet. 

 * Name: `grade_letter`
 * Argumente:
   * `score_points` - Die Note für eine Aufgabe in Punkten 
   * `total_points` - Die für die Aufgabe verfügbare Gesamtpunktzahl 
 * Rückgabe: 
   * Der Notenbuchstabe (z.B. "A", "B", "C")

```python
def grade_letter(score_points, total_points):
    # Berechnung Prozent
    percent = score_points/total_points*100
    print("percent: ", percent)
    if percent > 90:
        return "A"
    elif percent > 80:
        return "B"
    else:
        return "C"
grade_letter(50,100)
```

### 4. Plus und Minus

Verbessere deine Funktion `grade_letter_percent` nochmals, um `+` und `-` Noten einzubeziehen. Eine "+"-Note wird für die höchste prozentuale Note in der Kategorie vergeben, z.B. 99% ist eine "A+", 89% ist eine "B+", usw. Eine "-"-Note ist die niedrigste Note in der Kategorie, z.B. 90% ist eine A-, 80% ist eine B-, usw. 

 * Name: `grade_letter_plus`
 * Argumente:
   * `grade` - Die Note in Punkten 
   * `total` - Die Gesamtpunktzahl für die Aufgabe 
 * Rückgabe: 
   * Der Notenbuchstabe (z.B. "A+", "A", "A-" "B+")

```python
def grade_letter_plus(grade, total):
    # Berechnung Prozent
    percent = grade/total*100
    print("percent achieved: ", percent)
    # A
    if percent >= 99:
        return "A+"
    elif percent <99 and percent > 90:
        return "A"
    elif percent == 90:
        return "A-"
    # B
    elif percent == 89:
        return "B+"
    elif percent <89 and percent > 80:
        return "B"
    elif percent == 80:
        return "B-"
    # other
    else:
        return "C"
grade_letter_plus(89,100)
```
