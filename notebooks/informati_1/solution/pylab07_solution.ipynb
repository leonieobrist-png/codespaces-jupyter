### 6. Mileage

Write a program that computes how far your car can go if it gets 34.5 miles per gallon and has a 11.5 gallon tank.

````python
# 1. Solution
# direct:
print(34.5 * 11.5)
````

````python
# 2. Solution
# with variables and print statement as the output
miles_per_gallon = 34.5
gallon_tank = 11.5
# Here we use a so-called "format-string" (short: f-string), which is available since python 3.6.
# With the f-string, variables in the code can be accessed directly. 
# For comparison, the last line is an example of the old notation. \
# You do not need to know it. It is provided as additional information
print(f'Mit einem Tank von {gallon_tank} Gallonen und einer Reichweite von {miles_per_gallon} kann man {miles_per_gallon * gallon_tank} fahren.')
# 'older' notation
print('Mit einem Tank von', gallon_tank, 'Gallonen und einer Reichweite von', miles_per_gallon, 'kann man',  miles_per_gallon * gallon_tank, 'fahren.')
````

````python
# optionaml as function:
def Reichweite_Tank(verbrauch, tankgroesse):
    reichweite = verbrauch * tankgroesse
    return f'Mit einem Tank von {gallon_tank} Gallonen und einer Reichweite von {miles_per_gallon} kann man {reichweite} fahren.'
# Funktionsaufruf
Reichweite_Tank(miles_per_gallon, gallon_tank)
````

# 5. Variables and Logical Operators

So far you've learned about comparison operators like `>`, `<`, `==`, and `!=`. But what if you need to check multiple conditions at once? That's where **logical operators** come in!

Python has three logical operators:
- `and` - Returns `True` only if **both** conditions are true
- `or` - Returns `True` if **at least one** condition is true  
- `not` - **Flips** the result (True becomes False, False becomes True)

Think of them like this:
- `and` = "I need my keys AND my phone to leave the house"
- `or` = "I'll be happy if it's sunny OR if it's my birthday"
- `not` = "I'm NOT tired" (opposite of being tired)
