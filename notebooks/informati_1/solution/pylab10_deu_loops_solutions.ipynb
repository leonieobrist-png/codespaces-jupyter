## Teil 2: Die `for`-Loop und Wörterbücher 

Übe nun die Verwendung der `for`-Loops mit einem Dictionary. Hier ist ein Wörterbuch für den Anfang: 

```python
capital_city = {"Nepal": "Kathmandu", "Italy": "Rome", "England": "London", "Switzerland": "Bern"}
```

### 1. Loop über Dictionary-Keys

Schreibe eine "for"-Loop, die alle Schlüssel in `capital_city` ausgibt.

```python
# Solution 1
for key in capital_city:
    print(key)
```

```python
# Solution 2
for key in capital_city.keys():
    print(key)
```

### 2. Loop über Werte 

Schreibe eine `for`-Schleife, die alle Werte in `capital_city` ausgibt.

```python
# Solution 1
for key in capital_city:
    print(capital_city[key])
```

```python
# Solution 2
for value in capital_city.values():
    print(value)
```

### 3. Keys und Values 

Schreibe eine `for`-Schleife, die alle Schlüssel und Werte in `capital_city` ausgibt.

```python
# Solution 1
for key in capital_city:
    print(key, '-', capital_city[key])
```

```python
# solution 2
for land, hauptstadt in capital_city.items():
    print(f"{land} - {hauptstadt}")
```

## Teil 3: Der `while`-Loop

### 1. Zähler mit `while`- Loop
Verwende einen `while`-Loop, um die Zahlen von 1 bis 5 auszugeben.

```python
i = 1
while i <=5:
    print(i)
    i += 1
```

### 2. User Input
Schreibe ein Programm, dass dir 5 mal deinen Namen ausgibt. Verwende dafür die `input()` Funktion, um deinen Namen einzugeben.

```python
# mit user input
name = input('Name: ')
nr = 0 
while nr < 5:
    nr = nr +1
    print(name)
```

### 3. Über Liste iterieren  


Schreibe einen while-Loop, der jedes Element von `todo_list` ausgibt.

```python
index = 0
while index < len(todo_list):
    print(todo_list[index])
    index = index +1
```

### 4. Loop über Liste in umgekehrter Reihenfolge
Schreibe einen while-Loop, der jedes Element von `todo_list` **in umgekehrter Reihenfolge** ausgibt.

```python
index = len(todo_list) - 1
while index > 0:
    print(todo_list[index])
    index = index - 1
```

# Zusatz

### 4. Datei For-Schleife 

Schreibe eine for-Schleife, die jede Zeile der Datei `example.txt` **mit Zeilennummern** ausgibt.

```python
# Solution 1
file_handle = open('example.txt')

# initialize counter "nr"
nr = 0
for line in file_handle:
    nr = nr + 1
    print(nr, line.strip())
file_handle.close()
```

**Solution 2**

Mit dem "`with`-Argument muss eine Datei nicht mehr manuell geschlossen werden, wie in Solution 1.
`with` hält die Datei offen, nur so lange Python alle von euch spezifizierten Informationen aus der Datei gelesen hat. Danach wird die Datei automatisch geschlossen.

Die Methode `readlines()` erstellt euch eine Liste, deren Elemente jeweils eine Zeile aus der Text-Datei ist.

```python
# Zeilen aus Text-Datei lesen
with open("example.txt") as f:
    lines = f.readlines()
# ausgabe Liste namens "lines" 
print("Hier die generierte Liste 'lines':\n ", lines, "\n")

# nun kann man die Elemente der liste bearbeiten:
nr = 0
for line in lines:
    nr += 1
    print(nr, line.strip())
```

```python
# Solution 2.1: mit der "range()" methode
# Wir haben das "example.txt" zuvor importiert und als Liste gespeichert.
# Ein erneuter Import ist nicht mehr nötig.

# "range(len(lines))" gibt uns hier die Zahlenfolge 0,1,2,3,4

# print der "Zahlenfolge + 1" um es "menschen lesbar" zu machen => Zahlenfolge von 1 bis 5

# mit lines[i] greife ich auf die Elemente der Liste bei Index "i" zu, also lines[0], lines[1],..., lines[4]
# und kann wie gewohnt auf die einzelnen elemente der Liste, hier "strings (str)", Klassenfunktionen/Methoden anwenden.

for i in range(len(lines)):
    print(i+1, lines[i].strip())
```
