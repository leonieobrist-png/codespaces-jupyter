# Die `for`-Schleife

In dieser Lektion stelle ich eine einfache Syntax mit mächtigen Anwendungen vor. Zuvor haben Sie gelernt, wie man Listen erstellt und bearbeitet. Was aber, wenn Sie für *jedes* Element in einer Liste einen bestimmten Code ausführen möchten? Dazu benötigen Sie eine `for`-Schleife. Unten sehen Sie ein Beispiel:

Die `print`-Anweisung innerhalb der `for`-Schleife wird für jedes Element in der `animals`-Liste einmal ausgeführt. Geben Sie sie in die nächste Zelle ein:


````python
animals = ['Lions', 'Tigers', 'Bears'] 
for item in animals: 
    print(item) 
print('Oh my!')
````

Beachten Sie nochmals:

*   Die `print()`-Anweisung *innerhalb* der for-Schleife wird einmal für jedes Element in der Liste ausgeführt.
*   Die Elemente in der Liste werden nacheinander der Variablen `item` zugewiesen.


## Die `for`-Schleife

Dieses Bild zeigt die wichtigen Teile der `for`-Schleife:



Das Mächtigste an den `for`-Schleifen von Python ist die Idee, dass eine Sequenz rechts vom `in`-Schlüsselwort steht.

## Sequenzen

Eine Sequenz ist alles, was Element für Element durchlaufen werden kann. In Python können viele Datentypen wie Sequenzen behandelt werden. In einer früheren Lektion haben wir Listen kennengelernt, die von Natur aus Sequenzen sind, aber das ist nicht der erste Sequenztyp, den Sie kennengelernt haben. Die folgende Tabelle zeigt beispielhafte `for`-Schleifen mit Typen, die wir bereits gelernt haben.

| Typ | Sequenz von |
| --- | --- |
| `str` | Zeichen in der Zeichenkette |
| `list`| Elemente in der Liste |
| `dict` | Schlüssel im Wörterbuch |


Hier sind einige Beispiele:

### Ein String ist eine Sequenz von Buchstaben

Eine `for`-Schleife über einen String wird einmal pro Buchstabe im String ausgeführt:

````python
sentence = "Mary had a little lamb." 
for letter in sentence: 
    print(letter)
````

### Ein Dictionary ist eine Sequenz von Schlüsseln

Eine `for`-Schleife über ein Dictionary wird einmal pro Schlüssel ausgeführt:

````python
animals = { 'Lion': 'Panthera Leo', 
           'Tiger': 'Panthera Tigris', 
           'Bear': 'Ursus Arctos',
          }
for key in animals: 
    print("Der lateinische Name von", key, "ist", animals[key])
````

### Numerische Sequenzen

Wenn Sie schon einmal in einer anderen Sprache programmiert haben, haben Sie wahrscheinlich die `for`-Schleife im C-Stil gesehen. Wenn nicht, ist das kein Problem! In vielen anderen Sprachen dient die `for`-Schleife zum Zählen von Zahlen, und die Syntax ermöglicht es Ihnen, numerische Sequenzen zu erstellen (z.B. 0, 1, 2, ...), aber nicht, Dinge einfach zu "durchlaufen". Das Zählen von Zahlen ist manchmal wichtig, und Python bietet Ihnen eine Möglichkeit, das zu tun. Zum Beispiel:

```python 
for number in range(10):
    print(number)
```

Probieren Sie es in der nächsten Zelle aus:


**Welche Zahlen erzeugt die Liste?**

Die `range`-Funktion kann auch kompliziertere Sequenzen erzeugen. Wenn Sie `range` zwei Zahlen übergeben, beginnt sie bei der ersten und endet eine vor der zweiten. Zum Beispiel:

```python
for number in range(5, 10):
    print(number)
```

Versuchen Sie, die vorherige Zelle auf die Version von `range` mit zwei Argumenten zu ändern.




## Schleifenvariablen

Hier ist eine kleine Quizfrage! Was ist der Unterschied zwischen der Ausgabe dieser drei `for`-Schleifen:

```python
animals = ['Lions', 'Tigers', 'Bears']
for item in animals:
    print(item)

for animal in animals:
    print(animal)
    
for duck in animals:
    print(duck)
```

Nicht sicher? Probieren Sie sie aus:

**Antwort: NICHTS!** Der Unterschied in den `for`-Schleifen ist der Name der Schleifenvariable. Der Name der Schleifenvariable hat nichts Besonderes an sich, er ist einfach eine Variable, die die `for`-Schleife für Sie erstellt, um einzelne Elemente zu speichern. Es liegt an Ihnen, einen Namen zu wählen, der zu Ihrem Programm passt. Am besten wählen Sie einen Namen, der Ihnen hilft zu verstehen, was in der Variable enthalten ist. Hier ist eine bessere Frage:

Was ist der passendste Variablenname aus den obigen Schleifen?

1.  `item`
2.  `animal`
3.  `duck`
