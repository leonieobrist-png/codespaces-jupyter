# Übung zu Listen und Dictionaries mit JSON von openFDA

JSON steht in direktem Zusammenhang mit Listen und Dictionaries in Python, da es ähnliche Strukturen zur Darstellung von Daten verwendet. JSON-Objekte entsprechen Python-Dictionaries und JSON-Arrays entsprechen Python-Listen. Dies macht JSON zu einem idealen Format, um den Umgang mit Listen und Dictionaries zu üben.

In dieser Übung werden wir mit Daten von [openFDA](https://open.fda.gov/) arbeiten, einem Projekt, das öffentlichen Zugang zu hochwertigen Daten des öffentlichen Gesundheitswesens bietet. Dies ist besonders relevant für Studierende der biomedizinischen Diagnostik und Labortechnik, da es Einblicke in die Kennzeichnung von Medikamenten, unerwünschte Ereignisse und mehr bietet.

Durch die Verwendung der openFDA-API können Sie lernen, wie man komplexe verschachtelte Datenstrukturen parst, manipuliert und darauf zugreift.

## Was ist eine API?

API steht für Application Programming Interface (Anwendungsprogrammierschnittstelle). Es fungiert als Brücke zwischen verschiedenen Softwaresystemen und ermöglicht ihnen die Kommunikation untereinander. APIs definieren die Methoden und Datenformate, die Anwendungen zum Anfordern und Austauschen von Informationen verwenden können. Im Wesentlichen ermöglichen APIs, dass verschiedene Softwareprogramme einander verstehen und zusammenarbeiten.

## JSON: Ein gängiges Datenformat in APIs

JSON (JavaScript Object Notation) ist ein Datenformat, das Datenstrukturen in einer textbasierten Darstellung kapselt. Im Kern ist JSON eine Möglichkeit, Datenstrukturen so darzustellen, dass sie sowohl von Menschen als auch von Maschinen leicht lesbar sind. Die Einfachheit von JSON beruht auf der Verwendung von grundlegenden Datentypen und zwei fundamentalen Strukturen: Objekten (Dictionaries in Python) und Arrays (Listen in Python).

### Aufgabe 1: Arzneimitteldaten abrufen

1.  Machen Sie sich [hier](https://open.fda.gov/apis/drug/label/how-to-use-the-endpoint/) mit der Dokumentation der openFDA-API für Arzneimittelkennzeichnungen vertraut.
2.  Geben Sie einen Medikamentennamen (z. B. 'Aspirin', 'Ibuprofen') in die Variable `substance_name` ein.
3.  Führen Sie die folgende Zelle aus, um die Daten für das erste gefundene Medikament abzurufen.


```python
import requests

substance_name = 'Aspirin' # Geben Sie hier den Namen des Medikaments ein
limit = 10 # Maximum Anzahl Einträge, die aus der openFDA Datenbank zurückgegeben werden sollen

# URL für die API-Anfrage zusammenstellen
url = f'https://api.fda.gov/drug/label.json?search=openfda.substance_name:"{substance_name}"&limit={limit}'

# Anfrage an die openFDA-API senden
response = requests.get(url)

# Auf eine gültige Antwort prüfen
if response.status_code == 200:
    # JSON-Daten aus der Antwort parsen
    fda_data = response.json()
    print('Daten erfolgreich abgerufen!')
else:
    print(f'Datenabruf fehlgeschlagen: {response.status_code}')
    fda_data = {}
```
