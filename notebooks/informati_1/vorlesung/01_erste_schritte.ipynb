## Anweisungen / Statements

Eine *Anweisung* führt meist, aber nicht immer, eine Aktion aus, indem sie eine *Funktion* verwendet. Eine Funktion ist wie das Verb in einem Satz, sie beschreibt die Aktion, die ausgeführt wird. ### Funktionen 

Funktionen sind Teile eines Programms, die Eingaben (*Argumente*) entgegennehmen, Berechnungen durchführen und Ausgaben (*Return Values*) erzeugen. Die Argumente einer Funktion sind von Klammern --`(` und `)`-- umgeben und durch Kommas (`,`) getrennt. Funktionen können eine beliebige Anzahl von Argumenten haben, auch überhaupt keine. Der Rückgabewert kann auf die gleiche Weise wie jeder andere Wert verwendet werden. 

Die Funktion `len` gibt die Länge des Arguments zurück. 

```python
print(len("Hallo Welt"))
```

Funktionen wie `len` und `print` stehen für sich allein. Die meisten Funktionen sind jedoch *Klassenfunktionen*. Klassenfunktionen verwenden den Punkt (`.`), um das Subjekt der Funktion zu bestimmen. Man kann sich eine Klassenfunktion so vorstellen, als würde man einen englischen Satz in der Form Subjekt, Verb, Prädikat bilden. Hier ist ein Beispiel für eine Klassenfunktion:

```python 
"ich schreie nicht".upper()
```

Diese Anweisung kann man so lesen:
*Nimm die Buchstaben in "ich schreie nicht" und wandle sie in Großbuchstaben um.*

Von `""` umschlossene Ausdrücke werden als **string** (Zeichenkette) bezeichnet.

Ein weiteres Beispiel für Klassenfunktionen:

```python 
"ich schreie nicht".replace('nicht', 'wirklich')
```

Diese Anweisung kann wie folgt gelesen werden: 
*Nimm den Sting "ich schreie nicht" und ersetze "nicht" durch "wirklich".*


Ein großer Vorteil von Klassenfunktionen ist, dass sie manchmal miteinander verkettet werden können, um einen kompakten, aber lesbaren Code zu erstellen:

```python

```

### Zusammengesetzte Aussagen / Compound Statements

Einige Anweisungen können andere Anweisungen *umschließen*. Diese werden als *zusammengesetzte Anweisungen* bezeichnet. Das folgende Beispiel zeigt eine zusammengesetzte Anweisung, die zu verschiedenen Tageszeiten eine andere Begrüßung ausgibt. 

````python
from datetime import datetime 

# Get the current clock time.
now = datetime.now()  

# Compound statement: Decide what to say based on the current time
if now.hour < 12:
    print("Good morning.")
elif now.hour < 17:
    print("Good afternoon.")
else:
    print("Good evening.")
````

### Import Statements

Es gibt einen Grundbestand an Python-Funktionen, die immer verfügbar sind, darunter die `print`-Funktion. Um die volle Leistungsfähigkeit von Python zu nutzen, greifen Programme auf Zusatzfunktionen innerhalb von Packages zu. Jeder kann ein Python-Paket erstellen und es anderen zur Nutzung zur Verfügung stellen. Jupyter zum Beispiel ist ein Python-Paket. Die `import`-Anweisung macht Funktionen in einem Paket für das Programm verfügbar. 

