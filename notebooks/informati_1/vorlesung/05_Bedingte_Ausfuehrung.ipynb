# Logik und Entscheidungen

In dieser Lektion lernen wir, wie Programme Entscheidungen treffen und wie Sie Programme schreiben können, die auf Eingaben reagieren, indem sie einige Anweisungen ausführen und andere nicht.

## Entscheidungen mit `if` treffen

Was wäre, wenn Sie möchten, dass Ihr Code verschiedene Dinge basierend auf Benutzereingaben macht? Angenommen, Sie möchten ein Spiel spielen, das verschiedene Dinge basierend auf einer Wahl ausgibt. Die `if`-Anweisung ermöglicht es Ihnen, Python-Anweisungen bedingt auszuführen. Hier ist ein Diagramm einer `if`-Anweisung:



Die `if`-Anweisung führt Anweisungen im Körper der if-Anweisung bedingt aus. Der Code wird nur ausgeführt, wenn die Bedingung zu `True` ausgewertet wird. Mit anderen Worten, die `print`-Funktion im Bild wird möglicherweise ausgeführt oder nicht, je nach dem Wert von `number`.

Lassen Sie uns eine `if`-Anweisung verwenden, um uns zu sagen, wann wir die richtige Zahl erraten haben:


````python
number = 10 # Eingabe einer beliebigen Zahl

if number == 34: 
    print("That's right!")
````

**Beachten Sie, dass Sie "That's right!" nur sehen, wenn die Zahl im Widget 34 ist.**

Die *Bedingung* in der if-Anweisung ist `number == 34`. Wenn die Bedingung `True` ist, werden die Anweisungen im Körper des `if` ausgeführt, wenn die Bedingung `False` ist, werden die Anweisungen übersprungen. Sie können so viele Anweisungen haben, wie Sie möchten, innerhalb einer `if`-Anweisung, sogar andere `if`-Anweisungen.

### Die `if`-Bedingung

Die Anweisung, die dem `if`-Schlüsselwort folgt, wird zu `True` oder `False` ausgewertet. Während Sie programmieren, denken Sie daran, dass die Anweisung, die Sie schreiben, eine Frage in irgendeiner Form stellen muss. Im obigen Beispiel fragt der Code "wenn `number` gleich `34` ist?" Es gibt viele Formen, die eine Frage annehmen kann, aber sie haben alle eines gemeinsam: Sie werden zu einem `True`- oder `False`-Wert führen. Hier sind einige Beispiele für Fragen:

> Wenn `number` kleiner oder gleich `34` ist

```python
if number <= 34:
```

> Wenn `number` Modulo `2` gleich `0` ist (mit anderen Worten, wenn `number` gerade ist)?

```python
if (number % 2) == 0:
```

> Wenn `course` gleich dem String `"Python Programming for Everyone"` ist

```python
if course == "Python Programming for Everyone":
```

> Wenn `course` das Wort `Python` irgendwo darin hat

```python
if "Python" in course:
```

Während Sie an einem Programm arbeiten, denken Sie daran, dass Sie die bedingte Anweisung, die Sie in das `if` setzen, ausdrucken können, um zu sehen, ob Sie `True` oder `False` erhalten. Zum Beispiel:

```python
print(number <= 34)
```

Wenn das, was Sie erhalten, kein Boolean ist, bedeutet das wahrscheinlich, dass Sie irgendwo einen Fehler gemacht haben.


## `if` mit `else`

Die `else`-Anweisung lässt Sie Code ausführen, wenn etwas *nicht* wahr ist. Ein `else` kann nur in Kombination mit einem `if` verwendet werden. Zusammen lassen sie Sie eine Alternative ausführen, egal ob die Bedingung `True` oder `False` ist. Hier ist ein Diagramm der Struktur eines if/else-Paares.



Wenn Sie if/else haben, wird eine der Alternativen **immer** ausgeführt, je nach der Bedingung.

Beachten Sie, dass keine Bedingung dem `else` folgt. Was auch immer im `else`-Körper steht, wird nur ausgeführt, wenn die `if`-Bedingung `False` ist, und braucht daher keine weitere Information. Geben Sie das Beispiel in die nächste Zelle ein:


````python
number = 10 # Eingabe einer beliebigen Zahl
             
if number == 34:
    print("That's right!")
else:
    print("Guess again.")
````

## Mehrere Fragen mit `elif` stellen

In unserem Ratespiel gibt es wirklich drei Bedingungen, die uns interessieren:

1. Der Tipp ist richtig
1. Der Tipp ist zu niedrig
1. Der Tipp ist zu hoch

Mit `if` und `else` haben wir nur zwei Bedingungen erfasst: richtig und falsch. Es gibt mehr, was wir tun können. Wenn ein Programm auf eine Frage mit mehr als zwei Alternativen reagieren muss, ermöglicht uns die `elif`-Anweisung, eine weitere Frage zu stellen. Das `elif` ist kurz für "else if" und muss einer anfänglichen `if`-Anweisung folgen und wird nur ausgeführt, wenn die anfängliche Bedingung `False` ist.

Dieses Beispiel zeigt die beste Version unseres Spiels bisher:


````python
number = 10 # Eingabe einer beliebigen Zahl

if number == 34:
    print("That's right!")
elif number < 34:
    print("Too low!")
else:
    print("Too high!")
````

Hier sind die Regeln, die Sie sich merken sollten, wenn Sie `if`-, `elif`- und `else`-Anweisungen konstruieren:

1. Sie beginnen immer mit `if`
1. `elif` ist optional und Sie können so viele haben, wie Sie möchten.
1. `else` ist optional, aber es kann nur am Ende stehen.
1. In einer `if`-, `elif`-, `else`-Struktur wird **nur eine** Alternative ausgeführt.

Lassen Sie uns das Spielprogramm beenden, indem wir testen, ob der Benutzer eine Zahl zwischen 1 und 100 eingegeben hat, wie wir fragen:


````python
number = 10 # Eingabe einer beliebigen Zahl

if number == 34:
    print("That's right!")
elif number < 1 or number > 100:
    print("Bad guess.")
elif number < 34:
    print("Too low!")
else:
    print("Too high!")
````
