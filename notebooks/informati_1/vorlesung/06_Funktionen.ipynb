# Funktionen 

Eine *Funktion* gruppiert Anweisungen zusammen und gibt ihnen einen Namen. Sie sind der wichtigste Weg, um Code wiederzuverwenden. Python-Funktionen sind nach mathematischen Funktionen benannt und haben, genau wie ihre mathematischen Gegenstücke, Eingaben und Ausgaben. Funktionen beginnen mit dem `def`-Schlüsselwort. Unten ist ein Beispiel einer Funktion namens `print_tree`

Die `print_tree`-Funktion ist nur eine Sammlung von Anweisungen, die Sie bereits kennen. Die Anweisungen sind innerhalb der `def`-Anweisung eingerückt, um zu zeigen, dass sie zur Definition von `print_tree` gehören. Es gibt einen String oben, der *Docstring* genannt wird. Der Docstring beschreibt, was die Funktion macht.

Führen Sie nun den folgenden Code aus, was passiert?


````python
def print_tree():
    """Prints an ASCII tree"""
    print(" " * 3 + "*" * 1)
    print(" " * 2 + "*" * 3)
    print(" " * 1 + "*" * 5)
    print(" " * 0 + "*" * 7)
    print(" " * 2 + "|" * 3)
````

Der Code oben *definiert* die Funktion. Das Definieren einer Funktion macht den Namen der Funktion verfügbar, führt aber die Anweisungen *innerhalb* der Funktion nicht aus. Sie werden ausgeführt, wenn die Funktion *aufgerufen* oder invokiert wird. Dieser Code ruft `print_tree` auf:

```python
print_tree()
```

Rufen Sie `print_tree` in der nächsten Zelle auf:


````python
print_tree()
````

Funktionen können beliebig viele Argumente haben, null oder mehr. Die Namen der Argumente werden als Variablen innerhalb der Funktion verwendet. Hier ist ein Beispiel, warum Argumente so nützlich sind. Das ist die `triangle`-Funktion von vorhin mit einem Argument, das die Größe des Dreiecks steuert:


````python
def print_any_size_tree(size):
    """Display an ASCII tree of a given size"""
    # Tree leaves
    for i in range(size):
        spaces = " " * (size - i - 1)
        stars = "*" * (2 * i + 1)
        print(spaces + stars)
    # Tree trunk
    trunk_spaces = " " * (size - 2)
    print(trunk_spaces + "|" * 3)
````

## Funktionsvariablen und Gültigkeitsbereich 

Variablen, die innerhalb einer Funktion erstellt werden, einschließlich der Funktionsargumente, sind *privat* für die Funktion. Das bedeutet, sie existieren nur, während die Funktion ausgeführt wird, und verschwinden nach dem Rückgabewert der Funktion. Für Anfänger kann das verwirrend sein, aber es ist ein wesentliches Merkmal von Python und anderen Programmiersprachen. Die Gültigkeit einer Variable wird *Gültigkeitsbereich* genannt.


### Globale Variablen

Variablen, die ganz links definiert sind (auch Top-Level oder Dateibereich genannt), sind *globale Variablen*. Globale Variablen sind überall in einem Programm verfügbar, einschließlich innerhalb von Funktionen. Der Code unten zeigt, dass die Variable `a` innerhalb der Funktion verwendet werden kann:


````python
# a is a global variable.  
a = 10

def func(b):
    # b is private to the function and can 
    # only be used inside of it. 
    print('global (from outside) variable a =', 
          a, 'and private (form inside) variable b =',b)
    
# call func and pass in a+1
func(a+1)
````

Globale Variablen führen manchmal zu Verwirrung. Wenn eine Funktion eine Variable mit demselben Namen wie eine globale Variable verwendet, wird die innerhalb der Funktion verwendet. Hier ist ein Beispiel, wo ein Namenskonflikt zu verwirrenden Ergebnissen führt:


````python
a = 10 
print(a)

def func(a):
    a = 7
    print(a)

func(a)
print(a)
````

**Können Sie erklären, was passiert, wenn Sie diesen Code ausführen?** Denken Sie daran, es gibt *zwei* Variablen namens `a`.

```python

    """
    Calculate the power of a value x
    
    Arguments:
        x (float) - A number
        p (float) - The exponent 
        
    Returns: x to the power p.
    """
    return x**p
```


````python
def calculate_power_of(x, p):
    """
    Calculate the power of a value x

    Arguments:
        x (float) - A number
        p (float) - The exponent 

    Returns: x to the power p.
    """
    return x**p
````
