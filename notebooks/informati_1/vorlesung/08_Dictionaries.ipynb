# Dictionaries (Wörterbücher)

Dictionaries sind das Beste! Ihre Syntax ähnelt einer Liste und wie eine Liste enthält ein Dictionary eine beliebige Anzahl von Werten. Dictionaries sind jedoch anders, da sie anstelle von Werten an Indexpositionen *Schlüssel* (keys) verwenden. Stellen Sie sich vor, Sie möchten eine Sammlung wissenschaftlicher Tiernamen speichern. Mit anderen Worten, wenn Sie "Lion" in Ihr Suchwerkzeug eingeben, erhalten Sie als Ergebnis "Panthera Leo". Ein Dictionary ist genau das, was Sie brauchen. Stellen Sie sich ein Dictionary so vor:



In der Zeichnung ist der gebräuchliche Name eines Tieres der *Schlüssel* und der wissenschaftliche Name der *Wert*. In dieser Lektion lernen Sie, wie man Dictionaries erstellt, ändert und verwendet.

## Ein Dictionary erstellen

Dictionaries werden mit Schlüssel/Wert-Paaren deklariert, die durch Doppelpunkte getrennt sind. Hier ist ein Beispiel:

```python
animals = { 
    'Lion': 'Panthera Leo', 
    'Tiger': 'Panthera Tigris', 
    'Bear': 'Ursus Arctos',
    }
```

Im obigen Beispiel habe ich die Tiere auf verschiedene Zeilen aufgeteilt, um es lesbarer zu machen. Es ist genauso gültig, alles in einer Zeile zu schreiben:


````python
animals = {'Lion': 'Panthera Leo', 'Tiger': 'Panthera Tigris', 'Bear': 'Ursus Arctos'}
````

Genau wie eine Liste kann ein Dictionary leer sein. So erstellen Sie ein leeres Dictionary:

```python
animals = {} 
```

Dictionaries sind in Python ein eigener Typ. So verwenden Sie die `type`-Funktion, um herauszufinden, welchen Typ ein Dictionary hat:

```python
type(animals)
```

## Auf Dictionary-Elemente zugreifen, sie hinzufügen und entfernen

Auf Dictionaries wird genau wie auf Listen zugegriffen, aber mit Schlüsseln zwischen den `[` und `]`-Klammern. Zum Beispiel:

```python
print(animals['Lion'])
print(animals['Tiger'])
print(animals['Bear'])
```

Verwenden Sie die `print`-Funktion, um auf eines der Tiere im `animals`-Dictionary zuzugreifen:


Dictionary-Elemente können jederzeit erstellt werden. Zum Beispiel erstellen Sie hier ein Element, um den wissenschaftlichen Namen der gemeinen Hauskatze zu speichern:

```python
animals['Cat'] = 'Felis Silvestris Silvesteris'
```

Dictionary-Elemente werden neu zugewiesen, wenn sie links vom Gleichheitszeichen stehen:

```python
animals['Bear'] = 'Ursus Arctos Horribilis' # Grizzlybär
```


## Dictionary-Länge

Genau wie bei Listen gibt der `len`-Operator an, wie viele *Schlüssel/Wert-Paare* sich in einem Dictionary befinden.

```python
animals = { 'Lion': 'Panthera Leo', 'Tiger': 'Panthera Tigris', 'Bear': 'Ursus Arctos'}
print(len(animals))
```

## Suchen nach Schlüsseln mit `in`

Genau wie bei Listen durchsucht der `in`-Operator Dictionaries. Er sucht in den Schlüsseln nach einer Übereinstimmung. Sie können zum Beispiel überprüfen, ob Ihr `animals`-Dictionary einen Eintrag für 'Lion' enthält:

```python
animals = { 
    'Lion': 'Panthera Leo', 
    'Tiger': 'Panthera Tigris', 
    'Bear': 'Ursus Arctos'
}
'Lion' in animals
```

Sie können die Übereinstimmung mit `not` umkehren. Zum Beispiel:

```python
'Lion' not in animals
```

## Dictionary-Methoden: Listen aller Schlüssel oder Werte abrufen

Probieren das Beispiel unten aus, um die Schlüssel und Werte des Dictionaries zu erhalten:

````python
animals = { 
    'Lion': 'Panthera Leo', 
    'Tiger': 'Panthera Tigris', 
    'Bear': 'Ursus Arctos'
}
print(list(animals.keys()))
print(list(animals.values()))
````

## Key-Fehler

Wenn Sie auf einen Schlüssel zugreifen, der nicht existiert, erhalten Sie eine `KeyError`-Exception.

```python
print(animals['Sloth'])
```

Probieren Sie das Beispiel aus, um den Fehler zu sehen:


Eine Exception hält Ihr Programm an, es sei denn, Sie behandeln sie. Um einen Absturz zu vermeiden, ist es üblich, vor der Verwendung eines Schlüssels zu prüfen, ob er existiert, um sich gegen eine Exception abzusichern. Zum Beispiel:

```python
if 'Sloth' in animals:
    print ('Sloths are:', animals['Sloth'])
else:
    print ("Ich weiß nichts über Faultiere")
```


Dictionaries bieten die Funktion `get()`, die niemals eine Exception verursacht.

```python
print(animals.get('Sloth'))
```
