## Variable Names 

Variables can only have certain characters in their name to avoid confusion with Python syntax such as operators. There are a few rules to remember when you name your variables: 

1. A variable name must start with a letter or the underscore (`_`) character (no numbers).
1. A variable name can only contain alpha-numeric characters and underscores (A-Z, a-z, 0-9, and _ )
1. Variable names are case-sensitive (age, Age and AGE are three different variables)

The best thing to do is to name your variables descriptively, so that you know what they're used for. For example: 

```python
length = 100 
print(length)
```

Sometimes, it's handy to shorten variable names so you can go quicker:

```python
l = 100
print(l)
```

That can be confusing if it's not extremely obvious what the variable `l` is for. If in doubt, you can be as verbose as you like. The underscore `_` character means you can have very descriptive variable names. For example:

```python
length_of_side = 100
print(length_of_side)
```

What style you choose is up to you. Just remember, you are the programmer who is most likely to be confused by your code.

## Data Types 

Variables are how programs use the computer's RAM memory. Each variable is held in a location in memory and the location is associated with the *type* of data that's there. In many programming languages variables must only contain one type of value. For example, an integer or a string. In Python variables take any type of value depending on what they're assigned. Python is *dynamically typed*, meaning that variables can have any type and that type can change. That means a programmer has to be careful. Mixing variable types when performing operations is usually not allowed. The `type()` function shows you the type of a variable or literal: 

````python
print("The type of 10 is", type(10))
print("The type of 10.5 is", type(10.5))
print("The type of 'Python' is:", type('Python'))
````

Now let's update the program to use a variable instead of literals:

````python
a = 10
print("The type of a is", type(a))
a = 10.5
print("The type of a is", type(a))
a = 'Python'
print("The type of a is", type(a))
````

## The `input()` Function in Python

When programming, often there's a need to interact with the user. This is where the `input()` function in Python comes in. It allows a program to pause and wait for user input. When the user types something and presses enter, the 
`input()` function reads the user's input and can store it in a variable. This feature makes Python programs interactive. Note: The input windows appears on the top of your screen.

Here's how you can use the `input()` function:

````python
name = input("What is your name? ")
print("Hello, "+name+"!")
````

## Converting User Input for Calculations

In Python, the `input()` function reads everything as a string. If you expect the user to enter a number, you need to convert it using `int()` for integers or `float()` for floating-point numbers. Remember, Python is dynamically typed, so it's crucial to convert the user input to the correct type before performing any operations.

### Example: Calculating the Area of a Rectangle

Let's consider a simple example where we calculate the area of a rectangle. The area is found by multiplying the length by the width. Both of these values will be provided by the user.

````python
# Ask the user to enter the length and width of the rectangle
length = float(input("Enter the length of the rectangle: "))
width = float(input("Enter the width of the rectangle: "))

# Calculate the area
area = length * width

# Display the result
print("The area of the rectangle is:", area)
````

In this example, we use the `float()` function to convert the user's input into floating-point numbers. This is important because the user might enter decimal values. After the conversion, we can safely multiply `length` and `width` to get the `area`. 

Remember, failing to convert the inputs into numbers would result in a TypeError when you try to perform the multiplication, as Python can't multiply string types in this way.
