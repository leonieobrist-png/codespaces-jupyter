# Listen

Bisher haben die Variablen, mit denen wir gearbeitet haben, einen einzelnen Datenpunkt (eine ganze Zahl, eine Zeichenkette, etc.) gehalten, sogenannte *skalare* Variablen. Eine *Liste* in Python ist eine Variable, die eine beliebige Anzahl von Elementen enthalten kann. Die Werte werden *in Reihenfolge* gehalten. Denken Sie an meine To-Do-Liste:

1. Eat
2. Sleep
3. Program
4. Repeat

Eine Listenvariable gibt uns die Möglichkeit, alle vier Werte in einer einzigen Variablen zu speichern. Stellen Sie es sich so vor:



Beachten Sie, dass das erste Element in der Liste den **Index Null** hat. In dieser Lektion lernen Sie, wie man Listen erstellt, verwendet und ändert (oder *mutiert*).

## Auf Listenelemente zugreifen

Eine Listenvariable gibt Ihnen Zugriff auf einzelne Elemente in der Liste. Sie können Elemente wie jedes andere mit eckigen Klammern (`[` und `]`) verwenden. Wenn Sie eine Listenvariable verwenden, müssen Sie wählen, ob Sie die *gesamte Liste* oder nur *ein Element* verwenden möchten. Zum Beispiel:

```python
print(todo_list)      # Die ganze Liste ausgeben
print(todo_list[0])   # Das erste Element ausgeben
```

Probieren Sie beide print-Anweisungen in der nächsten Zelle aus:


**Listenindizes beginnen bei Null!** Element `[0]` der Liste ist das erste Element. Es ist etwas gewöhnungsbedürftig, aber es macht viele Dinge einfacher zu programmieren, wenn Listen bei Null beginnen.

<div class="alert alert-warning"><strong>Wichtig!</strong>

Bei Fragen rund um Listen ist die präzise Sprache wichtig (insbesondere im Prüfungskontext). Achten Sie auf den Unterschied zwischen:

- **"Index n"** bedeutet den genauen Index. Zum Beispiel:
    - "Gib Index drei von `todo_list` aus," bedeutet `print(todo_list[3])`.
- **"Das n-te Element von..."** bedeutet, dass *n* der Index plus eins ist. Zum Beispiel:
    - "Gib das dritte Element von `todo_list` aus," bedeutet `print(todo_list[2])`
</div>


Ein einzelnes Listenelement kann wie jede der bisher geschriebenen Variablen verwendet werden, die einen einzelnen Wert enthalten. Sie können auch Listenelemente zuweisen. Wenn Sie das tun, wird die Liste geändert. Hier ist ein Beispiel, wie Sie ein Listenelement ändern:


### Slicing

Eine Teilmenge einer Liste wird als *Slice* bezeichnet. Ein Slice kann mit dem Operator `[n:m]` erzeugt werden, der den Teil der Zeichenkette vom „n-ten“ Zeichen bis zum „m-ten“ Zeichen zurückgibt, **wobei das erste ausgeschlossen, aber das letzte eingeschlossen ist**.

```python
todo_list = ["Eat", "Sleep", "Program", "Repeat"]
print(todo_list[1:2])
print(todo_list[:2])
print(todo_list[1:])
print(todo_list[:])
```

Wenn Sie den ersten Index weglassen, beginnt der Slice am Anfang. Wenn Sie den zweiten weglassen,
geht der Slice bis zum Ende. Wenn Sie also beide weglassen, ist der Slice eine Kopie der gesamten Liste.

### Die `append` Funktion

Die `append` Funktion fügt ein Element am Ende der Liste hinzu. Zum Beispiel:

````python
todo_list = [] # Mit einer leeren Liste beginnen.
todo_list.append('Eat')
todo_list.append('Sleep')
todo_list.append('Program')
todo_list.append('Repeat')
print(todo_list)
````

### Die `insert` Funktion

Die `insert` Funktion fügt ein Element an einer beliebigen Stelle in die Liste ein. Der Aufruf von `insert` verschiebt vorhandene Elemente, die zum nächsten Index verschoben werden. Stellen Sie es sich so vor, dass sie sich nach rechts bewegen. 

**Können Sie die Reihenfolge nach Ausführung dieses Codes vorhersagen?** Versuchen Sie, ihn im Debugger auszuführen.

````python
todo_list = ['Eat', 'Repeat']
todo_list.insert(1, "Program") # Verschiebt 'Wiederholen'.
todo_list.insert(0, "Sleep") # Setzt dies an den Anfang und verschiebt alles andere.
print(todo_list)
````

## `split` und `join`: Konvertieren zwischen Listen und Zeichenketten

Es ist sehr üblich, eine Liste aus einer Zeichenkette zu erstellen und umgekehrt. Zum Beispiel haben Sie einen Satz und möchten ihn in einzelne Wörter zerlegen. Das geschieht mit der `split` Funktion. Hier ist ein Beispiel:

```python
text = "Mary goes to school."
words = text.split()
print(words)
```

Versuchen Sie `split` zu verwenden:


**Beachten Sie, dass der Punkt `.` immer noch da ist!** Das Gegenteil von `split` ist `join`. Die `join` Funktion nimmt eine Liste und fügt sie zu einer Zeichenkette zusammen. Die `join` Funktion sieht in Python etwas seltsam aus, da sie auf einem *Trennzeichen* operiert. Das Trennzeichen ist ein Zeichen, das zwischen die Listenelemente gesetzt wird. Hier ist das Gegenteil des `split` aus dem letzten Beispiel:

```python
words = ["Mary", "goes", "to", "School."]
text = " ".join(words)
print(text)
```

Beachten Sie das `" "`, das verwendet wird, um die Liste zusammenzufügen. Probieren Sie das Beispiel aus:
