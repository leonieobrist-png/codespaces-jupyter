# Die `while`-Schleife

Die `while`-Schleife ist als die *unbestimmte* Schleife bekannt. Sie wird verwendet, wenn man *nicht wissen kann*, wie oft man die Schleife durchlaufen muss. Die `while`-Schleife wird so lange wiederholt, wie die *Schleifenbedingung* `True` ist, und wird beendet, wenn die Bedingung `False` wird. Die `while`-Schleife ist seltener als die `for`-Schleife, aber dennoch für die meisten Programme unerlässlich.

Schauen wir uns unser Ratespiel noch einmal an. In dieser Version des Spiels möchten wir, dass der Benutzer so lange weiter rät, bis er die richtige Antwort findet. Es ist unmöglich zu wissen, wie viele Versuche er benötigen wird.


```python
guess = 0
while guess != 34:
    guess = int(input("Wählen Sie eine Zahl zwischen 1 und 100: "))
print("Sie haben es erraten!")
```

Der folgende Code kombiniert eine while-Schleife mit if-Anweisungen, um ein kleines Ratespiel mit einer Zufallszahl zu erstellen, die Sie nicht kennen. Probieren Sie es aus


```python
import random
guess = 0
random_number = random.randint(1,100)
while guess != random_number:
    guess = int(input("Wählen Sie eine Zahl zwischen 1 und 100: "))
    if guess < random_number:
        print('zu niedrig')
    elif guess > random_number:
        print('zu hoch')
print("Sie haben es erraten!")
```

## Eingabeüberprüfung

Die Schleife in der vorherigen Zelle fragt so lange nach einer Eingabe, bis der Benutzer den richtigen Wert angibt. Dies ist ein nützliches Verhalten, wenn Sie die Eingabe eines Benutzers *validieren* möchten. Hier ist ein Beispiel, ein Programm, das nach einer Zahl zwischen 1 und 100 fragt und den Benutzer zwingt, etwas Korrektes einzugeben.


```python
def input_number():
    number = -1
    while not 1 <= number <= 100:
        number = int(input("Geben Sie eine ganze Zahl zwischen 1 und 100 ein: "))
    return number

n = input_number()
print("Danke für die Eingabe von:", n)
```

## Endlosschleifen

`while`-Schleifen sind etwas gefährlicher als `for`-Schleifen, da ein Programmierfehler sie zu *Endlosschleifen* machen kann. In einer Endlosschleife stecken zu bleiben ist eine der Möglichkeiten, wie ein Programm abstürzen und nicht mehr reagieren kann. Daher ist bei der Programmierung einer `while`-Schleife mehr Sorgfalt geboten. Manchmal werden `while`-Schleifen als Endlosschleifen mit einer `break`- oder `return`-Anweisung geschrieben, die die Schleife beendet. Hier ist die Eingabeüberprüfungsfunktion erneut:


```python
def input_number():
    while True:
        number = int(input("Geben Sie eine ganze Zahl zwischen 1 und 100 ein: "))
        if 1 <= number <= 100:
            return number

n = input_number()
print("Danke für die Eingabe von:", n)
```
